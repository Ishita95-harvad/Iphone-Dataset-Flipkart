## Data loading

### Subtask:
Load the "Iphone Flipkart.csv" file into a pandas DataFrame.


**Reasoning**:
Load the "Iphone Flipkart.csv" file into a pandas DataFrame and display the first few rows.



```python
import pandas as pd

try:
    df = pd.read_csv('Iphone Flipkart.csv')
    display(df.head())
except FileNotFoundError:
    print("Error: 'Iphone Flipkart.csv' not found. Please ensure the file exists in the current directory or provide the correct path.")
    df = None
except Exception as e:
    print(f"An error occurred: {e}")
    df = None
```

## Data exploration

### Subtask:
Explore the loaded dataset to understand its characteristics.


**Reasoning**:
Explore the dataset by examining its shape, data types, missing values, descriptive statistics, unique values in categorical columns, and correlations between numerical features.



```python
# Data Shape
print("Data Shape:", df.shape)

# Data Types
print("\nData Types:\n", df.dtypes)

# Missing Values
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
print("\nMissing Values:\n", missing_values)
print("\nMissing Value Percentage:\n", missing_percentage)

# Descriptive Statistics for Numerical Columns
numerical_cols = ['xqddhh']  # Assuming 'xqddhh' represents ratings
try:
    numerical_stats = df[numerical_cols].describe()
    print("\nDescriptive Statistics for Numerical Columns:\n", numerical_stats)
except KeyError as e:
    print(f"Error: Column '{e}' not found in the DataFrame.")
    numerical_stats = None

# Unique Values for Categorical Columns
categorical_cols = ['Title', 'jigdf', 'jigdf2', 'jigdf3', 'jigdf4','jigdf5']
for col in categorical_cols:
    unique_values = df[col].nunique()
    print(f"\nUnique values in '{col}': {unique_values}")
    #print(df[col].value_counts())  # Uncomment to see frequency of each unique value if needed

# Correlation Analysis
try:
    correlation_matrix = df[['xqddhh']].corr() # Added a try-except block to handle potential errors, such as missing numerical columns
    print("\nCorrelation Matrix:\n", correlation_matrix)
except KeyError as e:
    print(f"\nError: Column '{e}' not found for correlation calculation.")
    correlation_matrix = None
```

## Data cleaning

### Subtask:
Clean the data by handling missing values and removing duplicate rows in the dataframe.


**Reasoning**:
Fill missing values in 'yray8j' and 'ukufwk' with their respective modes, remove duplicate rows, and then verify the cleaning process.



```python
# Fill missing values in 'yray8j' and 'ukufwk' with their mode
for col in ['yray8j', 'ukufwk']:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Verify data cleaning
missing_values_after = df.isnull().sum()
print("\nMissing Values After Cleaning:\n", missing_values_after)

duplicate_rows_after = df.duplicated().sum()
print("\nNumber of Duplicate Rows After Cleaning:", duplicate_rows_after)
```

## Data analysis

### Subtask:
Analyze the relationship between price, ratings, and reviews, and determine the distribution of iPhone models.


**Reasoning**:
Calculate descriptive statistics for price and ratings, analyze correlations, and determine the distribution of iPhone models.



```python
import pandas as pd
import numpy as np

# Identify columns for price, ratings, and reviews
price_col = 'yray8j'
ratings_col = 'xqddhh'
reviews_col = 'wphh3n1'  # Assuming 'wphh3n1' represents review counts

# Clean price column by removing ₹ and commas, and convert to numeric
df[price_col] = df[price_col].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)


# Descriptive statistics for price and ratings
price_stats = df[price_col].describe()
ratings_stats = df[ratings_col].describe()

print("Descriptive Statistics for Price:\n", price_stats)
print("\nDescriptive Statistics for Ratings:\n", ratings_stats)

# Correlation analysis
try:
    # Attempt Pearson correlation first
    correlation_price_ratings = df[[price_col, ratings_col]].corr(method='pearson')

    print("\nCorrelation between Price and Ratings (Pearson):\n", correlation_price_ratings)

    # If reviews column exists, analyze its correlation
    if reviews_col in df.columns:
        df[reviews_col] = df[reviews_col].astype(str).str.replace(',', '', regex=False).str.replace(' Reviews', '', regex=False).astype(float)
        correlation_price_reviews = df[[price_col, reviews_col]].corr(method='pearson')
        correlation_ratings_reviews = df[[ratings_col, reviews_col]].corr(method='pearson')
        print("\nCorrelation between Price and Reviews (Pearson):\n", correlation_price_reviews)
        print("\nCorrelation between Ratings and Reviews (Pearson):\n", correlation_ratings_reviews)
except TypeError as e:
    print(f"Error calculating correlation: {e}. Trying Spearman correlation instead.")
    correlation_price_ratings = df[[price_col, ratings_col]].corr(method='spearman')
    print("\nCorrelation between Price and Ratings (Spearman):\n", correlation_price_ratings)
    if reviews_col in df.columns:
        correlation_price_reviews = df[[price_col, reviews_col]].corr(method='spearman')
        correlation_ratings_reviews = df[[ratings_col, reviews_col]].corr(method='spearman')
        print("\nCorrelation between Price and Reviews (Spearman):\n", correlation_price_reviews)
        print("\nCorrelation between Ratings and Reviews (Spearman):\n", correlation_ratings_reviews)
except Exception as e:
    print(f"An unexpected error occurred: {e}")


# Distribution of iPhone models
model_counts = df['Title'].value_counts()
print("\nDistribution of iPhone models:\n", model_counts)

most_frequent_model = model_counts.index[0]
least_frequent_model = model_counts.index[-1]

print(f"\nMost frequent model: {most_frequent_model}")
print(f"\nLeast frequent model: {least_frequent_model}")
```

## Data visualization

### Subtask:
Visualize the key findings from the data analysis.


**Reasoning**:
Generate the visualizations as per the instructions, using the cleaned and analyzed data.



```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 15))

# 1. Histogram of iPhone prices
plt.subplot(2, 2, 1)
sns.histplot(df['yray8j'], kde=True, color='skyblue')
plt.title('Distribution of iPhone Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')

# 2. Scatter plot of price vs. ratings
plt.subplot(2, 2, 2)
sns.scatterplot(x='yray8j', y='xqddhh', data=df, color='coral')
plt.title('iPhone Price vs. Ratings')
plt.xlabel('Price')
plt.ylabel('Ratings')

# 3. Bar chart of iPhone model frequencies
plt.subplot(2, 2, 3)
model_counts = df['Title'].value_counts().sort_values(ascending=True)
sns.barplot(x=model_counts.index, y=model_counts.values, color='lightgreen')
plt.xticks(rotation=90)
plt.title('Frequency of iPhone Models')
plt.xlabel('iPhone Model')
plt.ylabel('Frequency')


# 4. Scatter plot of reviews vs. price (if 'wphh3n1' column exists)
if 'wphh3n1' in df.columns:
    plt.subplot(2, 2, 4)
    sns.scatterplot(x='wphh3n1', y='yray8j', data=df, color='plum')
    plt.title('Number of Reviews vs. Price')
    plt.xlabel('Number of Reviews')
    plt.ylabel('Price')

plt.tight_layout()
plt.show()
```
